```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Loss function (example: quadratic bowl)
def loss_fn(x, y):
    return x**2 + y**2  # Simple quadratic loss function

# 2. Gradient function (∇L)
def grad_fn(x, y):
    return 2 * x, 2 * y  # Gradients with respect to x and y (derivatives)

# 3. SGD algorithm (optimization path)
def run_sgd(start_x, start_y, lr=0.1, steps=30):
    path = [(start_x, start_y)]  # Initialize the path with the starting point
    x, y = start_x, start_y
    for _ in range(steps):
        dx, dy = grad_fn(x, y)  # Calculate the gradient at the current position
        x -= lr * dx  # Update x by subtracting the gradient scaled by the learning rate
        y -= lr * dy  # Update y by subtracting the gradient scaled by the learning rate
        path.append((x, y))  # Append the updated position to the path
    return np.array(path)

# 4. Create grid for surface and path visualization
path = run_sgd(start_x=2.0, start_y=1.5)  # Run SGD starting from (2.0, 1.5)
x = np.linspace(-3, 3, 100)  # Generate x values for the grid
y = np.linspace(-3, 3, 100)  # Generate y values for the grid
X, Y = np.meshgrid(x, y)  # Create a meshgrid for the 2D surface
Z = loss_fn(X, Y)  # Calculate the loss values for each point on the grid

# 5. Static visualization
plt.figure(figsize=(8, 6))  # Create a figure with a specific size
plt.contourf(X, Y, Z, levels=50, cmap='viridis')  # Plot filled contour for the loss surface
plt.colorbar()  # Add a colorbar to indicate the loss values
plt.plot(path[:, 0], path[:, 1], 'ro-', markersize=5)  # Plot the entire path of optimization
plt.scatter(path[0, 0], path[0, 1], color='blue', label='Start')  # Plot the start point in blue
plt.scatter(path[-1, 0], path[-1, 1], color='gold', s=100, marker='*', label='End')  # Plot the end point in gold (with larger size and star shape)
plt.title("2D Gradient Descent (SGD)")  # Title for the plot
plt.xlabel("x")  # Label for the x-axis
plt.ylabel("y")  # Label for the y-axis
plt.legend()  # Display the legend to differentiate start and end points
plt.show()  # Display the plot
```
