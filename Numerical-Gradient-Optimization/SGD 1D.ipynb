```python
import numpy as np
import matplotlib.pyplot as plt

## Target function L(θ)=(θ−3)^2

# Minimum point is 3

def loss(theta):
    return (theta-3) ** 2

def grad(theta):
    return 2 * (theta-3)

# Initial Value and learning Rate

theta = 0.0
lr = 0.1
thetas = [theta]
losses = [loss(theta)]

# SGD steps

for i in range(30):
    g = grad(theta)
    theta = theta - lr * g 
    thetas.append(theta)
    losses.append(loss(theta))

#Visilization
x = np.linspace(-1,7,100)
y = loss(x)

plt.plot(x,y,label="Loss Function")
plt.scatter(thetas,losses,c='red',label='SGD Steps')
plt.title("SGD Optimization of (θ - 3)²")
plt.xlabel("θ")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```
